```python
import random
import time
import matplotlib.pyplot as plt



def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        mini = i
        for j in range(i, n):
            if arr[j] < arr[mini]:
                mini = j
        arr[i], arr[mini] = arr[mini], arr[i]
```

```python

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
```

```python
import sys
def Merge(arr,low,mid,high):
    n1=mid-low+1
    n2=high-mid
    
    L=[0]*n1
    R=[0]*n2
    
    for i in range(n1):
        L[i]=arr[low+i-1]
    for i in range(n2):
        R[i]=arr[mid+1+i]
        
    L.append(sys.maxsize)
    R.append(sys.maxsize)
    
    i, j=0, 0
    for k in range(low,high):
        if(L[i]<=R[j]):
            arr[k]=L[i]
            i=i+1
        else:
            arr[k]=R[j]
            j=j+1


def merge_sort(arr,low,high):
    if low<high:
        mid=(low + high)//2
    
        merge_sort(arr,low,mid)
        merge_sort(arr,mid+1,high)
        Merge(arr,low,mid,high)
```

```python
def Quick_Sort(arr,low,high):
    if low<high:
        mid=partition(arr,low,high)
        Quick_Sort(arr,low,mid-1)
        Quick_Sort(arr,mid+1,high)

def partition(arr,low,high):
    x=arr[high]
    i=low-1
    for j in range(low,high):
        if arr[j]<= x:
            i=i+1
            arr[i],arr[j]=arr[j],arr[i]
    arr[i+1],arr[high]=arr[high],arr[i+1]
    return i+1
```

```python
input_sizes = [1000,5000,10000,15000]  # you can increase this, but sorting will slow
bubble_times = []
selection_times = []
insertion_times = []
Merge_times=[]
Quick_times=[]
```

```python
for size in input_sizes:
    arr = [random.randint(1, 100000) for _ in range(size)]

    # Bubble Sort
    arr_copy = arr.copy()
    start = time.perf_counter()
    bubble_sort(arr_copy)
    end = time.perf_counter()
    bubble_time = end - start
    bubble_times.append(bubble_time)

    # Selection Sort
    arr_copy = arr.copy()
    start = time.perf_counter()
    selection_sort(arr_copy)
    end = time.perf_counter()
    selection_time = end - start
    selection_times.append(selection_time)

    # Insertion Sort
    arr_copy = arr.copy()
    start = time.perf_counter()
    insertion_sort(arr_copy)
    end = time.perf_counter()
    insertion_time = end - start
    insertion_times.append(insertion_time)
    
    #Merge Sort
    arr_copy=arr.copy()
    start=time.perf_counter()
    merge_sort(arr_copy,0,(len(arr_copy)-1))
    end=time.perf_counter()
    MergeSort_time=end-start
    Merge_times.append(MergeSort_time)
    
    #Insertion Sort
    arr_copy=arr.copy()
    start=time.perf_counter()
    Quick_Sort(arr,0,(len(arr_copy)-1))
    end=time.perf_counter()
    Quick_time=end-start
    Quick_times.append(Quick_time)
    

    print(f"Input size: {size}")
    print(f"  Bubble Sort time: {bubble_time:.6f} seconds")
    print(f"  Selection Sort time: {selection_time:.6f} seconds")
    print(f"  Insertion Sort time: {insertion_time:.6f} seconds")
    print(f"  Merge Sort time: {MergeSort_time:.6f} seconds")
    print(f"  Quick Sort time: {Quick_time:.6f} seconds\n")

    
```
