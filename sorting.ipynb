```python
import random
import time
import matplotlib.pyplot as plt



def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        mini = i
        for j in range(i, n):
            if arr[j] < arr[mini]:
                mini = j
        arr[i], arr[mini] = arr[mini], arr[i]
```

```python

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
```

```python
input_sizes = [1000,5000,10000,15000,20000,25000]  # you can increase this, but sorting will slow
bubble_times = []
selection_times = []
insertion_times = []
```

```python
for size in input_sizes:
    arr = [random.randint(1, 100000) for _ in range(size)]

    # Bubble Sort
    arr_copy = arr.copy()
    start = time.perf_counter()
    bubble_sort(arr_copy)
    end = time.perf_counter()
    bubble_time = end - start
    bubble_times.append(bubble_time)

    # Selection Sort
    arr_copy = arr.copy()
    start = time.perf_counter()
    selection_sort(arr_copy)
    end = time.perf_counter()
    selection_time = end - start
    selection_times.append(selection_time)

    # Insertion Sort
    arr_copy = arr.copy()
    start = time.perf_counter()
    insertion_sort(arr_copy)
    end = time.perf_counter()
    insertion_time = end - start
    insertion_times.append(insertion_time)

    print(f"Input size: {size}")
    print(f"  Bubble Sort time: {bubble_time:.6f} seconds")
    print(f"  Selection Sort time: {selection_time:.6f} seconds")
    print(f"  Insertion Sort time: {insertion_time:.6f} seconds\n")
```

```python
# Plotting
plt.figure(figsize=(6,4))
plt.plot(input_sizes, bubble_times, label='Bubble Sort', marker='o')
plt.plot(input_sizes, selection_times, label='Selection Sort', marker='s')
plt.plot(input_sizes, insertion_times, label='Insertion Sort', marker='^')

plt.title('Sorting Algorithm Time Complexity')
plt.xlabel('Input Size (Number of elements)')
plt.ylabel('Time Taken (seconds)')
plt.legend()
plt.grid(True)
plt.show()
```
